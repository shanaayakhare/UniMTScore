```python
from bert_score import score

print("BERTScore library ready ✅")
print("Languages in our evaluation:", val_semantic["language"].unique().tolist())
print("Total samples:", len(val_semantic))
```

```python
translations = val_semantic["translation"].tolist()
references = val_semantic["ref"].tolist()

P, R, F1 = score(
    translations,
    references,
    lang="hi",
    verbose=True
)

val_semantic["bertscore_f1"] = F1.numpy()

print("BERTScore calculated successfully ✅")
print(val_semantic[["language", "bertscore_f1"]].head())
```

```python
print("BERTScore results ready ✅")
print("\nColumns:")
print(val_semantic.columns.tolist())

print("\nFirst 5 BERTScore values:")
print(val_semantic[["language", "bertscore_f1"]].head())

print("\nBERTScore statistics:")
print(val_semantic["bertscore_f1"].describe())
```

```python
print("BERTScore results ready ✅")

print("\nBERTScore statistics:")
print(val_semantic["bertscore_f1"].describe())

print("\nLanguage-wise mean BERTScore:")
print(
    val_semantic.groupby("language")["bertscore_f1"]
    .mean()
    .sort_values(ascending=False)
)
```

```python
language_codes = {
    "Hindi": "hi",
    "Gujarati": "gu",
    "Marathi": "mr",
    "Malayalam": "ml"
}

val_semantic["bertscore_f1_final"] = 0.0

for language, code in language_codes.items():
    mask = val_semantic["language"] == language

    translations = val_semantic.loc[mask, "translation"].tolist()
    references = val_semantic.loc[mask, "ref"].tolist()

    print(f"\nCalculating BERTScore for {language}...")

    P, R, F1 = score(
        translations,
        references,
        lang=code,
        verbose=True
    )

    val_semantic.loc[mask, "bertscore_f1_final"] = F1.numpy()

print("\nFinal multilingual BERTScore calculation completed ✅")
```

```python
from scipy.stats import pearsonr

# Overall Pearson correlation
mqm_corr, mqm_p = pearsonr(
    val_semantic["bertscore_f1_final"],
    val_semantic["mqm_norm_score"]
)

da_corr, da_p = pearsonr(
    val_semantic["bertscore_f1_final"],
    val_semantic["da_norm_score"]
)

print("BERTScore vs MQM")
print(f"Pearson correlation: {mqm_corr:.4f}")
print(f"P-value: {mqm_p:.4f}")

print("\nBERTScore vs DA")
print(f"Pearson correlation: {da_corr:.4f}")
print(f"P-value: {da_p:.4f}")
```

```python
for language in required_languages:
    data = val_semantic[val_semantic["language"] == language]

    mqm_r, _ = pearsonr(
        data["bertscore_f1_final"],
        data["mqm_norm_score"]
    )

    da_r, _ = pearsonr(
        data["bertscore_f1_final"],
        data["da_norm_score"]
    )

    print(f"{language}:")
    print(f"  BERTScore vs MQM: {mqm_r:.4f}")
    print(f"  BERTScore vs DA : {da_r:.4f}")
    print()
```

```python
print("Final BERTScore validation ✅")

print("Missing BERTScore values:",
      val_semantic["bertscore_f1_final"].isnull().sum())

print("Minimum BERTScore:",
      val_semantic["bertscore_f1_final"].min())

print("Maximum BERTScore:",
      val_semantic["bertscore_f1_final"].max())

print("Total samples:",
      len(val_semantic))
```
