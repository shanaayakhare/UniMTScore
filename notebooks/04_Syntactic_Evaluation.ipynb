```python
from difflib import SequenceMatcher

def extract_pos_dep(text, pipeline):
    doc = pipeline(text)

    pos_tags = []
    dep_tags = []

    for sentence in doc.sentences:
        for word in sentence.words:
            pos_tags.append(word.upos)
            dep_tags.append(word.deprel)

    return pos_tags, dep_tags
```

```python
ref_pos, ref_dep = extract_pos_dep(
    row["reference"],
    pipelines["Marathi"]
)

mt_pos, mt_dep = extract_pos_dep(
    row["translation"],
    pipelines["Marathi"]
)

print("Reference POS:", ref_pos)
print("Translation POS:", mt_pos)

print("\nReference Dependencies:", ref_dep)
print("Translation Dependencies:", mt_dep)
```

```python
def sequence_similarity(seq1, seq2):
    return SequenceMatcher(None, seq1, seq2).ratio()

pos_similarity = sequence_similarity(ref_pos, mt_pos)

print("POS Similarity:", pos_similarity)
```

```python
from collections import Counter

def dependency_similarity(dep1, dep2):
    if not dep1 and not dep2:
        return 1.0
    if not dep1 or not dep2:
        return 0.0

    c1 = Counter(dep1)
    c2 = Counter(dep2)

    overlap = sum((c1 & c2).values())

    precision = overlap / sum(c2.values())
    recall = overlap / sum(c1.values())

    if precision + recall == 0:
        return 0.0

    return 2 * precision * recall / (precision + recall)
```

```python
dep_similarity = dependency_similarity(ref_dep, mt_dep)

print("Dependency Similarity:", dep_similarity)
```

```python
syntactic_score = 0.5 * pos_similarity + 0.5 * dep_similarity

print("POS Similarity:", pos_similarity)
print("Dependency Similarity:", dep_similarity)
print("Unified Syntactic Score:", syntactic_score)
```

```python
def compute_syntactic_scores(row):
    language = row["language"]

    if language not in pipelines:
        return {
            "pos_similarity": None,
            "dependency_similarity": None,
            "syntactic_score": None
        }

    pipeline = pipelines[language]

    ref_pos, ref_dep = extract_pos_dep(
        row["reference"],
        pipeline
    )

    mt_pos, mt_dep = extract_pos_dep(
        row["translation"],
        pipeline
    )

    pos_score = sequence_similarity(
        ref_pos,
        mt_pos
    )

    dep_score = dependency_similarity(
        ref_dep,
        mt_dep
    )

    syntactic_score = 0.5 * pos_score + 0.5 * dep_score

    return {
        "pos_similarity": pos_score,
        "dependency_similarity": dep_score,
        "syntactic_score": syntactic_score
    }
```

```python
sample = val[
    val["language"].isin(["Hindi", "Marathi", "Tamil"])
].head(10).copy()

results = sample.apply(
    compute_syntactic_scores,
    axis=1
)

sample["pos_similarity"] = results.apply(
    lambda x: x["pos_similarity"]
)

sample["dependency_similarity"] = results.apply(
    lambda x: x["dependency_similarity"]
)

sample["syntactic_score"] = results.apply(
    lambda x: x["syntactic_score"]
)

display(
    sample[
        [
            "language",
            "reference",
            "translation",
            "pos_similarity",
            "dependency_similarity",
            "syntactic_score",
            "mqm_norm_score",
            "da_norm_score"
        ]
    ]
)
```

```python
supported_languages = list(pipelines.keys())

val_supported = val[
    val["language"].isin(supported_languages)
].copy()

print("Rows to process:", len(val_supported))
```

```python
from tqdm.auto import tqdm
tqdm.pandas()

results = val_supported.progress_apply(
    compute_syntactic_scores,
    axis=1
)

val_supported["pos_similarity"] = results.apply(
    lambda x: x["pos_similarity"]
)

val_supported["dependency_similarity"] = results.apply(
    lambda x: x["dependency_similarity"]
)

val_supported["syntactic_score"] = results.apply(
    lambda x: x["syntactic_score"]
)
```

```python
from scipy.stats import pearsonr, spearmanr

clean = val_supported.dropna(
    subset=["syntactic_score", "mqm_norm_score", "da_norm_score"]
)

print("Rows used:", len(clean))

pearson_mqm = pearsonr(
    clean["syntactic_score"],
    clean["mqm_norm_score"]
)

pearson_da = pearsonr(
    clean["syntactic_score"],
    clean["da_norm_score"]
)

spearman_mqm = spearmanr(
    clean["syntactic_score"],
    clean["mqm_norm_score"]
)

spearman_da = spearmanr(
    clean["syntactic_score"],
    clean["da_norm_score"]
)

print("\nPearson correlation:")
print("Syntactic vs MQM:", pearson_mqm.statistic)
print("Syntactic vs DA :", pearson_da.statistic)

print("\nSpearman correlation:")
print("Syntactic vs MQM:", spearman_mqm.statistic)
print("Syntactic vs DA :", spearman_da.statistic)
```

```python
for lang, group in clean.groupby("language"):
    if len(group) < 3:
        continue

    p_mqm = group["syntactic_score"].corr(
        group["mqm_norm_score"],
        method="pearson"
    )

    s_mqm = group["syntactic_score"].corr(
        group["mqm_norm_score"],
        method="spearman"
    )

    print(
        lang,
        "| Pearson MQM:", round(p_mqm, 4),
        "| Spearman MQM:", round(s_mqm, 4)
    )
```

```python
def compute_syntactic_scores(row):
    language = row["language"]

    supported = ["Hindi", "Marathi", "Tamil", "Gujarati"]

    if language not in supported:
        return {
            "pos_similarity": None,
            "dependency_similarity": None,
            "syntactic_score": None
        }

    ref_pos, ref_dep = extract_pos_dep_by_language(
        row["reference"],
        language
    )

    mt_pos, mt_dep = extract_pos_dep_by_language(
        row["translation"],
        language
    )

    pos_score = sequence_similarity(ref_pos, mt_pos)
    dep_score = dependency_similarity(ref_dep, mt_dep)

    syntactic_score = 0.5 * pos_score + 0.5 * dep_score

    return {
        "pos_similarity": pos_score,
        "dependency_similarity": dep_score,
        "syntactic_score": syntactic_score
    }
```

```python
from difflib import SequenceMatcher
from collections import Counter

def sequence_similarity(seq1, seq2):
    return SequenceMatcher(None, seq1, seq2).ratio()

def dependency_similarity(dep1, dep2):
    if not dep1 and not dep2:
        return 1.0
    if not dep1 or not dep2:
        return 0.0

    c1 = Counter(dep1)
    c2 = Counter(dep2)

    overlap = sum((c1 & c2).values())

    precision = overlap / sum(c2.values())
    recall = overlap / sum(c1.values())

    if precision + recall == 0:
        return 0.0

    return 2 * precision * recall / (precision + recall)

print("Helper functions loaded.")
```

```python
from difflib import SequenceMatcher
from collections import Counter

def sequence_similarity(seq1, seq2):
    return SequenceMatcher(None, seq1, seq2).ratio()

def dependency_similarity(dep1, dep2):
    if not dep1 and not dep2:
        return 1.0

    if not dep1 or not dep2:
        return 0.0

    c1 = Counter(dep1)
    c2 = Counter(dep2)

    overlap = sum((c1 & c2).values())

    precision = overlap / sum(c2.values())
    recall = overlap / sum(c1.values())

    if precision + recall == 0:
        return 0.0

    return 2 * precision * recall / (precision + recall)
```

```python
def extract_pos_dep_by_language(text, language):
    text = str(text)

    if language == "Gujarati":
        doc = gu_pipeline([text.split()])

    elif language == "Malayalam":
        doc = ml_pipeline([text.split()])

    else:
        doc = pipelines[language](text)

    pos_tags = []
    dep_tags = []

    for sentence in doc.sentences:
        for word in sentence.words:
            pos_tags.append(word.upos)
            dep_tags.append(word.deprel)

    return pos_tags, dep_tags
```

```python
def compute_syntactic_scores(row):
    target_language = row["target_language"]

    ref_pos, ref_dep = extract_pos_dep_by_language(
        row["reference"],
        target_language
    )

    mt_pos, mt_dep = extract_pos_dep_by_language(
        row["translation"],
        target_language
    )

    pos_score = sequence_similarity(ref_pos, mt_pos)
    dep_score = dependency_similarity(ref_dep, mt_dep)

    syntactic_score = (
        0.5 * pos_score
        + 0.5 * dep_score
    )

    return {
        "pos_similarity": pos_score,
        "dependency_similarity": dep_score,
        "syntactic_score": syntactic_score
    }
```

```python
import numpy as np
import pandas as pd
from scipy.stats import pearsonr, spearmanr, kendalltau

weight_results = []

# POS weight: 0.0, 0.1, ..., 1.0
for pos_weight in np.arange(0.0, 1.01, 0.1):

    dep_weight = 1.0 - pos_weight

    temp_score = (
        pos_weight * val["pos_similarity"]
        + dep_weight * val["dependency_similarity"]
    )

    for human in ["mqm_norm_score", "da_norm_score"]:

        temp = pd.DataFrame({
            "syntax": temp_score,
            "human": val[human]
        }).dropna()

        pearson = pearsonr(
            temp["syntax"],
            temp["human"]
        )[0]

        spearman = spearmanr(
            temp["syntax"],
            temp["human"]
        )[0]

        kendall = kendalltau(
            temp["syntax"],
            temp["human"]
        )[0]

        weight_results.append({
            "pos_weight": round(pos_weight, 1),
            "dependency_weight": round(dep_weight, 1),
            "human_score": human,
            "pearson": pearson,
            "spearman": spearman,
            "kendall": kendall
        })

weight_results = pd.DataFrame(weight_results)

display(
    weight_results.round(4)
)
```

```python
best_weights = (
    weight_results
    .sort_values(
        ["human_score", "spearman"],
        ascending=[True, False]
    )
    .groupby("human_score")
    .head(1)
)

display(best_weights.round(4))
```

```python
weight_summary = (
    weight_results
    .groupby(
        ["pos_weight", "dependency_weight"]
    )[["pearson", "spearman", "kendall"]]
    .mean()
    .reset_index()
)

weight_summary["overall_correlation"] = (
    weight_summary["pearson"]
    + weight_summary["spearman"]
    + weight_summary["kendall"]
) / 3

weight_summary = weight_summary.sort_values(
    "overall_correlation",
    ascending=False
)

display(weight_summary.round(4))
```

```python
best_row = weight_summary.iloc[0]

BEST_POS_WEIGHT = float(best_row["pos_weight"])
BEST_DEP_WEIGHT = float(best_row["dependency_weight"])

print("Selected POS weight:", BEST_POS_WEIGHT)
print("Selected Dependency weight:", BEST_DEP_WEIGHT)
print("Validation overall correlation:", best_row["overall_correlation"])
```

```python
from collections import Counter

def extract_dependency_labels(text, language):
    text = str(text)

    if language == "Gujarati":
        doc = gu_pipeline([text.split()])

    elif language == "Malayalam":
        doc = ml_pipeline([text.split()])

    elif language == "Odia":
        doc = odia_pipeline(text)

    else:
        doc = pipelines[language](text)

    labels = []

    for sentence in doc.sentences:
        for word in sentence.words:
            labels.append(word.deprel)

    return labels


def dependency_similarity(ref_labels, mt_labels):
    if not ref_labels and not mt_labels:
        return 1.0

    if not ref_labels or not mt_labels:
        return 0.0

    ref_counter = Counter(ref_labels)
    mt_counter = Counter(mt_labels)

    overlap = sum((ref_counter & mt_counter).values())

    precision = overlap / sum(mt_counter.values())
    recall = overlap / sum(ref_counter.values())

    if precision + recall == 0:
        return 0.0

    return 2 * precision * recall / (precision + recall)


def compute_dependency_score(row):
    language = row["target_language"]

    ref_labels = extract_dependency_labels(
        row["reference"],
        language
    )

    mt_labels = extract_dependency_labels(
        row["translation"],
        language
    )

    return dependency_similarity(
        ref_labels,
        mt_labels
    )

print("Dependency scoring functions ready.")
```

```python
from difflib import SequenceMatcher

# ==========================================
# 1. POS TAG EXTRACTION
# ==========================================

def extract_pos_tags(text, language):

    text = str(text)

    if language == "Gujarati":
        doc = gu_pipeline([text.split()])

    elif language == "Malayalam":
        doc = ml_pipeline([text.split()])

    elif language == "Odia":
        doc = odia_pipeline(text)

    else:
        doc = pipelines[language](text)

    pos_tags = []

    for sentence in doc.sentences:
        for word in sentence.words:
            pos_tags.append(word.upos)

    return pos_tags


# ==========================================
# 2. POS SEQUENCE SIMILARITY
# ==========================================

def pos_similarity(ref_tags, mt_tags):

    return SequenceMatcher(
        None,
        ref_tags,
        mt_tags
    ).ratio()


# ==========================================
# 3. ROW-LEVEL POS SIMILARITY
# ==========================================

def compute_pos_similarity(row):

    target_language = row["target_language"]

    reference_tags = extract_pos_tags(
        row["reference"],
        target_language
    )

    translation_tags = extract_pos_tags(
        row["translation"],
        target_language
    )

    return pos_similarity(
        reference_tags,
        translation_tags
    )


# ==========================================
# 4. FINAL CHECK
# ==========================================

print("extract_pos_tags:", callable(extract_pos_tags))
print("compute_pos_similarity:", callable(compute_pos_similarity))

print("\n✅ POS functions restored.")
print("Rows ready for scoring:", len(pos_pending_df))
```

```python
from pathlib import Path

SAVE_DIR = Path("/content/kannada_ud")

def count_conllu_sentences(path):
    text = path.read_text(encoding="utf-8")
    sentences = [
        block for block in text.strip().split("\n\n")
        if block.strip()
    ]
    return len(sentences)

print("===== KANNADA DATASET =====")

total = 0

for split in ["train", "dev", "test"]:
    path = SAVE_DIR / f"{split}.conllu"

    if not path.exists():
        print(f"❌ {split}: FILE NOT FOUND")
        continue

    n = count_conllu_sentences(path)
    total += n

    print(
        f"✅ {split:<5} | "
        f"{n:,} sentences | "
        f"{path.stat().st_size / 1024 / 1024:.2f} MB"
    )

print("-" * 45)
print(f"Total: {total:,} sentences")
```

```python
from collections import Counter

def extract_kannada_dependency_labels(text):
    # Basic whitespace tokenization
    tokens = str(text).strip().split()

    if not tokens:
        return []

    prediction = kannada_parser.predict(
        [tokens],
        lang=None,
        prob=False,
        verbose=False
    )

    sentence = prediction.sentences[0]

    # SuPar CoNLL output:
    # values[7] = predicted DEPREL labels
    return list(sentence.values[7])


def dependency_similarity(ref_labels, mt_labels):
    if not ref_labels and not mt_labels:
        return 1.0

    if not ref_labels or not mt_labels:
        return 0.0

    ref_counter = Counter(ref_labels)
    mt_counter = Counter(mt_labels)

    overlap = sum((ref_counter & mt_counter).values())

    precision = overlap / sum(mt_counter.values())
    recall = overlap / sum(ref_counter.values())

    if precision + recall == 0:
        return 0.0

    return 2 * precision * recall / (precision + recall)


# ----- TEST -----

reference = "ನಾನು ಇಂದು ಶಾಲೆಗೆ ಹೋಗುತ್ತೇನೆ ."
translation = "ನಾನು ಶಾಲೆಗೆ ಇಂದು ಹೋಗುತ್ತೇನೆ ."

ref_labels = extract_kannada_dependency_labels(reference)
mt_labels = extract_kannada_dependency_labels(translation)

score = dependency_similarity(ref_labels, mt_labels)

print("Reference DEPREL:", ref_labels)
print("MT DEPREL       :", mt_labels)
print("\nDependency similarity:", score)
```

```python
# ===== PREPARE KANNADA POS DATA =====

def load_kannada_pos_data(conllu_path):
    sentences = []
    tags = []

    current_tokens = []
    current_tags = []

    with open(conllu_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()

            # End of sentence
            if not line:
                if current_tokens:
                    sentences.append(current_tokens)
                    tags.append(current_tags)

                current_tokens = []
                current_tags = []
                continue

            if line.startswith("#"):
                continue

            parts = line.split("\t")

            if (
                len(parts) >= 10
                and "-" not in parts[0]
                and "." not in parts[0]
            ):
                token = parts[1]
                upos = parts[3]

                if upos != "_":
                    upos = POS_NORMALIZATION.get(upos, upos)

                    current_tokens.append(token)
                    current_tags.append(upos)

    # Last sentence if file doesn't end with blank line
    if current_tokens:
        sentences.append(current_tokens)
        tags.append(current_tags)

    return sentences, tags


train_tokens, train_tags = load_kannada_pos_data(
    "/content/kannada_ud/train.conllu"
)

dev_tokens, dev_tags = load_kannada_pos_data(
    "/content/kannada_ud/dev.conllu"
)

test_tokens, test_tags = load_kannada_pos_data(
    "/content/kannada_ud/test.conllu"
)


print("===== KANNADA POS DATA READY =====")

print("Train:", len(train_tokens))
print("Dev  :", len(dev_tokens))
print("Test :", len(test_tokens))

print("\nExample tokens:")
print(train_tokens[0])

print("\nExample UPOS:")
print(train_tags[0])

print("\nLengths match:",
      len(train_tokens[0]) == len(train_tags[0]))
```

```python
# ===== CREATE POS LABEL MAPPING =====

all_labels = sorted(
    set(tag for sentence_tags in train_tags for tag in sentence_tags)
)

label2id = {label: i for i, label in enumerate(all_labels)}
id2label = {i: label for label, i in label2id.items()}

print("===== KANNADA POS LABEL MAPPING =====")
print("Number of labels:", len(all_labels))

for label, idx in label2id.items():
    print(f"{idx:2d} -> {label}")

# Basic verification
assert len(label2id) == 17
assert len(label2id) == len(id2label)

print("\n✅ Label mapping ready")
```

```python
# ===== LOAD MBERT TOKENIZER + TEST POS ALIGNMENT =====

from transformers import AutoTokenizer

MODEL_NAME = "bert-base-multilingual-cased"

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    use_fast=True
)

# Take one Kannada training sentence
sample_words = train_tokens[0]
sample_tags = train_tags[0]

encoding = tokenizer(
    sample_words,
    is_split_into_words=True,
    truncation=True,
    max_length=256
)

word_ids = encoding.word_ids()

# Align POS labels:
# only first subword of each original word gets the POS label
aligned_labels = []
previous_word_id = None

for word_id in word_ids:

    # [CLS], [SEP], etc.
    if word_id is None:
        aligned_labels.append(-100)

    # First subword of a word
    elif word_id != previous_word_id:
        aligned_labels.append(
            label2id[sample_tags[word_id]]
        )

    # Remaining subwords ignored in loss
    else:
        aligned_labels.append(-100)

    previous_word_id = word_id


tokens = tokenizer.convert_ids_to_tokens(
    encoding["input_ids"]
)

print("===== MBERT ALIGNMENT TEST =====")

for token, word_id, label_id in zip(
    tokens,
    word_ids,
    aligned_labels
):
    if label_id == -100:
        label = "IGNORE"
    else:
        label = id2label[label_id]

    print(
        f"{token:20s} | "
        f"word_id={str(word_id):4s} | "
        f"label={label}"
    )

print("\nOriginal words :", len(sample_words))
print(
    "Assigned labels:",
    sum(x != -100 for x in aligned_labels)
)

assert (
    sum(x != -100 for x in aligned_labels)
    == len(sample_words)
)

print("\n✅ Kannada word ↔ mBERT subword alignment correct")
```

```python
# ===== TOKENIZE FULL KANNADA POS DATASETS =====

def tokenize_and_align_batch(sentences, tag_sequences):
    encodings = tokenizer(
        sentences,
        is_split_into_words=True,
        truncation=True,
        max_length=256,
        padding=False
    )

    all_aligned_labels = []

    for i, tags in enumerate(tag_sequences):
        word_ids = encodings.word_ids(batch_index=i)

        aligned_labels = []
        previous_word_id = None

        for word_id in word_ids:

            if word_id is None:
                aligned_labels.append(-100)

            elif word_id != previous_word_id:
                aligned_labels.append(
                    label2id[tags[word_id]]
                )

            else:
                aligned_labels.append(-100)

            previous_word_id = word_id

        all_aligned_labels.append(aligned_labels)

    encodings["labels"] = all_aligned_labels
    return encodings


print("===== TOKENIZING KANNADA POS DATA =====")

train_encodings = tokenize_and_align_batch(
    train_tokens,
    train_tags
)
print("✅ Train tokenized:", len(train_encodings["input_ids"]))

dev_encodings = tokenize_and_align_batch(
    dev_tokens,
    dev_tags
)
print("✅ Dev tokenized:", len(dev_encodings["input_ids"]))

test_encodings = tokenize_and_align_batch(
    test_tokens,
    test_tags
)
print("✅ Test tokenized:", len(test_encodings["input_ids"]))


# Verification
assert len(train_encodings["input_ids"]) == len(train_tokens)
assert len(dev_encodings["input_ids"]) == len(dev_tokens)
assert len(test_encodings["input_ids"]) == len(test_tokens)

print("\nExample encoded length:",
      len(train_encodings["input_ids"][0]))

print("Example label length:",
      len(train_encodings["labels"][0]))

assert (
    len(train_encodings["input_ids"][0])
    ==
    len(train_encodings["labels"][0])
)

print("\n✅ FULL KANNADA POS TOKENIZATION COMPLETE")
```

```python
# ===== CREATE PYTORCH DATASETS FOR KANNADA POS =====

import torch
from torch.utils.data import Dataset

class KannadaPOSDataset(Dataset):
    def __init__(self, encodings):
        self.encodings = encodings

    def __len__(self):
        return len(self.encodings["input_ids"])

    def __getitem__(self, idx):
        return {
            key: torch.tensor(value[idx])
            for key, value in self.encodings.items()
        }


train_dataset = KannadaPOSDataset(train_encodings)
dev_dataset   = KannadaPOSDataset(dev_encodings)
test_dataset  = KannadaPOSDataset(test_encodings)


print("===== PYTORCH DATASETS READY =====")

print("Train:", len(train_dataset))
print("Dev  :", len(dev_dataset))
print("Test :", len(test_dataset))


# Inspect one sample
sample = train_dataset[0]

print("\nSample fields:")
for key, value in sample.items():
    print(
        f"{key:15s}",
        "shape =", tuple(value.shape)
    )


# Verification
assert len(train_dataset) == len(train_tokens)
assert len(dev_dataset) == len(dev_tokens)
assert len(test_dataset) == len(test_tokens)

assert (
    len(sample["input_ids"])
    ==
    len(sample["labels"])
)

print("\n✅ KANNADA POS PYTORCH DATASETS READY")
```

```python
# ===== INITIALIZE MBERT KANNADA POS MODEL =====

import os
import torch
from transformers import AutoModelForTokenClassification

MODEL_NAME = "bert-base-multilingual-cased"

POS_MODEL_DIR = (
    "/content/drive/MyDrive/"
    "MT_Evaluation_Project/models/kannada_pos"
)

os.makedirs(POS_MODEL_DIR, exist_ok=True)

print("===== INITIALIZING KANNADA POS MODEL =====")

pos_model = AutoModelForTokenClassification.from_pretrained(
    MODEL_NAME,
    num_labels=len(label2id),
    id2label=id2label,
    label2id=label2id
)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

pos_model.to(device)

print("\nBase model :", MODEL_NAME)
print("POS labels :", len(label2id))
print("Device     :", device)
print("Save path  :", POS_MODEL_DIR)

print(
    "Parameters :",
    f"{sum(p.numel() for p in pos_model.parameters()):,}"
)

print("\n✅ KANNADA MBERT POS MODEL INITIALIZED")
print("⚠️ Training has NOT started yet.")
```

```python
# ===== CONFIGURE KANNADA POS TRAINING =====

import numpy as np
from transformers import (
    TrainingArguments,
    Trainer,
    DataCollatorForTokenClassification
)

# Dynamic padding for token classification
data_collator = DataCollatorForTokenClassification(
    tokenizer=tokenizer
)

training_args = TrainingArguments(
    output_dir=POS_MODEL_DIR,

    # Training
    num_train_epochs=5,
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    weight_decay=0.01,

    # Evaluation + checkpoint
    eval_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    greater_is_better=False,

    # GPU
    fp16=torch.cuda.is_available(),

    # Keep Drive storage under control
    save_total_limit=2,

    logging_steps=100,
    report_to="none"
)

trainer = Trainer(
    model=pos_model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=dev_dataset,
    data_collator=data_collator
)

print("===== KANNADA POS TRAINER READY =====")
print("Train sentences :", len(train_dataset))
print("Dev sentences   :", len(dev_dataset))
print("Epochs          :", training_args.num_train_epochs)
print("Learning rate   :", training_args.learning_rate)
print("Train batch     :", training_args.per_device_train_batch_size)
print("FP16            :", training_args.fp16)
print("Device          :", device)

print("\n✅ TRAINER CONFIGURED")
print("⚠️ Training has NOT started yet.")
```

```python
# ===== TRAIN KANNADA MBERT POS TAGGER =====

print("===== STARTING KANNADA POS TRAINING =====")
print("⚠️ Do not disconnect/restart the runtime during training.\n")

train_result = trainer.train()

print("\n===== TRAINING FINISHED =====")
print("Training loss :", train_result.training_loss)

# Save BEST model loaded by Trainer
FINAL_POS_MODEL_DIR = (
    "/content/drive/MyDrive/"
    "MT_Evaluation_Project/models/kannada_pos/final"
)

trainer.save_model(FINAL_POS_MODEL_DIR)
tokenizer.save_pretrained(FINAL_POS_MODEL_DIR)

print("\n✅ KANNADA POS TRAINING COMPLETE")
print("✅ Best model saved to:")
print(FINAL_POS_MODEL_DIR)
```

```python
# ===== KANNADA POS INFERENCE FUNCTION =====

import torch

def extract_kannada_pos_tags(text):
    words = str(text).strip().split()

    if not words:
        return []

    encoding = tokenizer(
        words,
        is_split_into_words=True,
        return_tensors="pt",
        truncation=True,
        max_length=256
    )

    word_ids = encoding.word_ids(batch_index=0)

    encoding = {
        key: value.to(device)
        for key, value in encoding.items()
    }

    pos_model.eval()

    with torch.no_grad():
        outputs = pos_model(**encoding)

    predictions = outputs.logits.argmax(dim=-1)[0].cpu().tolist()

    # Keep prediction only for first subword of each original word
    tags = []
    previous_word_id = None

    for word_id, pred_id in zip(word_ids, predictions):

        if word_id is not None and word_id != previous_word_id:
            tags.append(id2label[pred_id])

        previous_word_id = word_id

    return tags


# ----- TEST -----

test_text = "ನಾನು ಇಂದು ಶಾಲೆಗೆ ಹೋಗುತ್ತೇನೆ ."

tags = extract_kannada_pos_tags(test_text)

print("===== KANNADA POS PREDICTION =====")

print("Words:")
print(test_text.split())

print("\nPredicted UPOS:")
print(tags)

print("\nWord → POS")
for word, tag in zip(test_text.split(), tags):
    print(f"{word:20s} -> {tag}")

assert len(test_text.split()) == len(tags)

print("\n✅ KANNADA POS EXTRACTION WORKING")
```

```python
# ===== FULL KANNADA SYNTACTIC SCORING — 3,524 ROWS =====

import os
import pandas as pd
from tqdm.auto import tqdm

KANNADA_CHECKPOINT = (
    "/content/drive/MyDrive/"
    "MT_Evaluation_Project/results/"
    "kannada_syntactic_scores_partial.csv"
)

os.makedirs(os.path.dirname(KANNADA_CHECKPOINT), exist_ok=True)

# --------------------------------------------------
# LOAD EXISTING CHECKPOINT
# --------------------------------------------------

if os.path.exists(KANNADA_CHECKPOINT):

    completed_df = pd.read_csv(KANNADA_CHECKPOINT)

    completed_ids = set(
        completed_df["evaluation_id"].astype(str)
    )

    print("Existing checkpoint found.")
    print("Already completed:", len(completed_df))

else:

    completed_df = pd.DataFrame()

    completed_ids = set()

    print("No checkpoint found.")
    print("Starting Kannada scoring from beginning.")


# --------------------------------------------------
# FIND ONLY PENDING ROWS
# --------------------------------------------------

pending_df = kannada_df[
    ~kannada_df["evaluation_id"]
    .astype(str)
    .isin(completed_ids)
].copy()

print("\n===== KANNADA SCORING STATUS =====")
print("Total Kannada rows :", len(kannada_df))
print("Completed          :", len(completed_ids))
print("Pending            :", len(pending_df))


# --------------------------------------------------
# SCORE ONLY PENDING ROWS
# --------------------------------------------------

new_results = []

for count, (_, row) in enumerate(
    tqdm(
        pending_df.iterrows(),
        total=len(pending_df),
        desc="Kannada syntactic scoring"
    ),
    start=1
):

    try:

        reference = str(row["reference"])
        translation = str(row["translation"])

        # ---------- POS ----------

        ref_pos = extract_kannada_pos_tags(reference)
        mt_pos = extract_kannada_pos_tags(translation)

        pos_score = pos_similarity(
            ref_pos,
            mt_pos
        )

        # ---------- DEPENDENCY ----------

        ref_dep = extract_kannada_dependency_labels(reference)
        mt_dep = extract_kannada_dependency_labels(translation)

        dep_score = dependency_similarity(
            ref_dep,
            mt_dep
        )

        # ---------- COMBINED ----------

        syntactic_score = (
            0.5 * pos_score
            +
            0.5 * dep_score
        )

        result = row.to_dict()

        result["pos_similarity"] = pos_score
        result["dependency_similarity"] = dep_score
        result["syntactic_score"] = syntactic_score
        result["error"] = ""

        new_results.append(result)

    except Exception as e:

        result = row.to_dict()

        result["pos_similarity"] = None
        result["dependency_similarity"] = None
        result["syntactic_score"] = None

        result["error"] = (
            type(e).__name__ + ": " + str(e)
        )

        new_results.append(result)


    # --------------------------------------------------
    # SAVE EVERY 25 ROWS
    # --------------------------------------------------

    if count % 25 == 0:

        batch_df = pd.DataFrame(new_results)

        if len(completed_df):
            save_df = pd.concat(
                [completed_df, batch_df],
                ignore_index=True
            )
        else:
            save_df = batch_df

        save_df = save_df.drop_duplicates(
            subset="evaluation_id",
            keep="last"
        )

        save_df.to_csv(
            KANNADA_CHECKPOINT,
            index=False
        )

        print(
            f"\n💾 Checkpoint saved: "
            f"{len(save_df)}/3524"
        )


# --------------------------------------------------
# FINAL SAVE
# --------------------------------------------------

batch_df = pd.DataFrame(new_results)

if len(completed_df):
    final_df = pd.concat(
        [completed_df, batch_df],
        ignore_index=True
    )
else:
    final_df = batch_df

final_df = final_df.drop_duplicates(
    subset="evaluation_id",
    keep="last"
)

final_df.to_csv(
    KANNADA_CHECKPOINT,
    index=False
)


# --------------------------------------------------
# FINAL STATUS
# --------------------------------------------------

successful = final_df[
    final_df["syntactic_score"].notna()
]

failed = final_df[
    final_df["syntactic_score"].isna()
]

print("\n===== KANNADA FINAL STATUS =====")

print("Total saved :", len(final_df))
print("Successful  :", len(successful))
print("Failed      :", len(failed))

print(
    "Duplicates  :",
    final_df["evaluation_id"].duplicated().sum()
)

print("\nCheckpoint:")
print(KANNADA_CHECKPOINT)

if (
    len(final_df) == 3524
    and len(successful) == 3524
    and len(failed) == 0
    and final_df["evaluation_id"].duplicated().sum() == 0
):
    print("\n🎉 KANNADA SYNTACTIC SCORING COMPLETE")
else:
    print("\n⚠️ Some rows still need attention.")
```

```python
# ===== VALIDATION — CURRENT 50/50 SYNTACTIC CORRELATION =====

import pandas as pd
from scipy.stats import pearsonr, spearmanr, kendalltau

VALIDATION_PATH = (
    "/content/drive/MyDrive/"
    "MT_Evaluation_Project/results/"
    "validation_syntactic_scores.csv"
)

val_df = pd.read_csv(VALIDATION_PATH)

print("===== VALIDATION DATA =====")
print("Rows:", len(val_df))

# Current methodology
val_df["syntactic_score"] = (
    0.5 * val_df["pos_similarity"]
    +
    0.5 * val_df["dependency_similarity"]
)

print("\nFormula:")
print("Syntactic = 0.5 × POS + 0.5 × Dependency")


def calculate_correlations(df, human_col):

    valid = df[
        ["syntactic_score", human_col]
    ].dropna()

    pearson = pearsonr(
        valid["syntactic_score"],
        valid[human_col]
    )[0]

    spearman = spearmanr(
        valid["syntactic_score"],
        valid[human_col]
    )[0]

    kendall = kendalltau(
        valid["syntactic_score"],
        valid[human_col]
    )[0]

    return pearson, spearman, kendall, len(valid)


results = []

for human_col in [
    "mqm_norm_score",
    "da_norm_score"
]:

    p, s, k, n = calculate_correlations(
        val_df,
        human_col
    )

    results.append({
        "human_metric": human_col,
        "n": n,
        "pearson": p,
        "spearman": s,
        "kendall": k
    })


correlation_df = pd.DataFrame(results)

print("\n===== OVERALL HUMAN CORRELATION =====")

print(
    correlation_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

print("\n✅ CURRENT 50/50 SYNTACTIC CORRELATION CALCULATED")
```
