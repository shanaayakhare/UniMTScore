```python
from comet.encoders.xlmr import XLMREncoder

def patched_forward(
    self, input_ids, attention_mask, **kwargs
):
    outputs = self.model(
        input_ids=input_ids,
        attention_mask=attention_mask,
        output_hidden_states=True,
        return_dict=False,
    )

    last_hidden_states = outputs[0]
    all_layers = outputs[-1]

    return {
        "sentemb": last_hidden_states[:, 0, :],
        "wordemb": last_hidden_states,
        "all_layers": all_layers,
        "attention_mask": attention_mask,
    }

XLMREncoder.forward = patched_forward

print("COMET XLM-R compatibility patch applied ✅")
```

```python
from comet import download_model, load_from_checkpoint

model_path = download_model("Unbabel/wmt22-comet-da")
model = load_from_checkpoint(model_path)

print("COMET model loaded successfully ✅")
```

```python
from comet.encoders.xlmr import XLMREncoder

def patched_forward(
    self, input_ids, attention_mask, **kwargs
):
    outputs = self.model(
        input_ids=input_ids,
        attention_mask=attention_mask,
        output_hidden_states=True,
        return_dict=False,
    )

    last_hidden_states = outputs[0]
    all_layers = outputs[-1]

    return {
        "sentemb": last_hidden_states[:, 0, :],
        "wordemb": last_hidden_states,
        "all_layers": all_layers,
        "attention_mask": attention_mask,
    }

XLMREncoder.forward = patched_forward

print("COMET XLM-R compatibility patch applied ✅")
```

```python
comet_output = model.predict(
    comet_data,
    batch_size=4,
    gpus=0
)

val_semantic["comet_score"] = comet_output.scores

COMET_PATH = "/content/drive/MyDrive/Machine Translation Project/IndicMT-Eval/IndicMT_Eval_Project/comet_scores_800.csv"

val_semantic.to_csv(COMET_PATH, index=False)

print("COMET scoring completed ✅")
print("Number of scores:", len(comet_output.scores))
print("Scores saved to Drive ✅")
print("Saved at:", COMET_PATH)
```

```python
from scipy.stats import pearsonr

print("Overall COMET Correlations")
print("-" * 35)

mqm_corr, mqm_p = pearsonr(
    val_semantic["comet_score"],
    val_semantic["mqm_norm_score"]
)

da_corr, da_p = pearsonr(
    val_semantic["comet_score"],
    val_semantic["da_norm_score"]
)

print(f"COMET vs MQM: r = {mqm_corr:.4f}, p = {mqm_p:.4g}")
print(f"COMET vs DA : r = {da_corr:.4f}, p = {da_p:.4g}")
```

```python
languages = ["Hindi", "Gujarati", "Marathi", "Malayalam"]

print("Language-wise COMET Correlations")
print("=" * 50)

for language in languages:
    subset = val_semantic[val_semantic["language"] == language]

    mqm_r, mqm_p = pearsonr(
        subset["comet_score"],
        subset["mqm_norm_score"]
    )

    da_r, da_p = pearsonr(
        subset["comet_score"],
        subset["da_norm_score"]
    )

    print(f"\n{language}")
    print(f"  COMET vs MQM: r = {mqm_r:.4f}, p = {mqm_p:.4g}")
    print(f"  COMET vs DA : r = {da_r:.4f}, p = {da_p:.4g}")
```

```python
comparison_data = []

for language in languages:
    subset = val_semantic[val_semantic["language"] == language]

    bert_mqm, _ = pearsonr(
        subset["bertscore_f1_final"],
        subset["mqm_norm_score"]
    )

    bert_da, _ = pearsonr(
        subset["bertscore_f1_final"],
        subset["da_norm_score"]
    )

    comet_mqm, _ = pearsonr(
        subset["comet_score"],
        subset["mqm_norm_score"]
    )

    comet_da, _ = pearsonr(
        subset["comet_score"],
        subset["da_norm_score"]
    )

    comparison_data.append({
        "Language": language,
        "BERTScore-MQM": bert_mqm,
        "BERTScore-DA": bert_da,
        "COMET-MQM": comet_mqm,
        "COMET-DA": comet_da
    })

comparison_df = pd.DataFrame(comparison_data)

print(comparison_df.round(4).to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import numpy as np

languages = comparison_df["Language"]

x = np.arange(len(languages))
width = 0.2

plt.figure(figsize=(10, 6))

plt.bar(x - 1.5*width, comparison_df["BERTScore-MQM"], width, label="BERTScore vs MQM")
plt.bar(x - 0.5*width, comparison_df["BERTScore-DA"], width, label="BERTScore vs DA")
plt.bar(x + 0.5*width, comparison_df["COMET-MQM"], width, label="COMET vs MQM")
plt.bar(x + 1.5*width, comparison_df["COMET-DA"], width, label="COMET vs DA")

plt.xlabel("Language")
plt.ylabel("Pearson Correlation (r)")
plt.title("Language-wise Semantic Metric Correlation")
plt.xticks(x, languages)
plt.ylim(0, 0.75)
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
semantic_correlation_results = []

for language in languages:
    subset = val_semantic[val_semantic["language"] == language]

    for metric in ["bertscore_f1_final", "comet_score"]:
        for human_score in ["mqm_norm_score", "da_norm_score"]:

            r, p = pearsonr(
                subset[metric],
                subset[human_score]
            )

            semantic_correlation_results.append({
                "language": language,
                "metric": metric,
                "human_score": human_score,
                "n": len(subset),
                "pearson": r,
                "pearson_p": p
            })

semantic_correlation_results = pd.DataFrame(
    semantic_correlation_results
)

print("Semantic language-wise correlation results:")
display(semantic_correlation_results.round(4))
```
